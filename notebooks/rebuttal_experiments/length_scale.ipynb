```python
from src.data.data import *
from src.embedor import *
from src.plotting import *
import matplotlib
import seaborn as sns
import umap
import numpy as np
from sklearn.manifold import TSNE, Isomap, SpectralEmbedding, MDS
import phate
import os

%load_ext autoreload

output_dir = 'figures'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
```

```python
def create_gaussian_mixture(n_samples, dim):
    """Create a Gaussian mixture dataset where the density is drastically different in the two components."""
    sigma_1 = 0.1
    sigma_3 = 0.5
    sigma_2 = (sigma_1 + sigma_3)/3
    mu_1 = np.zeros(dim)
    mu_3 = np.ones(dim)
    mu_3 /= np.linalg.norm(mu_3)
    mu_3 *= 2.12

    mu_2 = (mu_1 + mu_3) / 3
    X_1 = np.random.normal(mu_1, sigma_1, (n_samples // 3, dim))
    X_2 = np.random.normal(mu_2, sigma_2, (n_samples // 3, dim))
    X_3 = np.random.normal(mu_3, sigma_3, (n_samples // 3, dim))
    X = np.vstack((X_1, X_2, X_3))
    labels = np.array([0] * (n_samples // 3) + [1] * (n_samples // 3) + [2] * (n_samples // 3))
    return X, labels

def create_gaussian(mu, sigma, n_samples, dim):
    """Create a Gaussian dataset."""
    return np.random.normal(mu, sigma, (n_samples, dim))

dim = 2
n_samples = 3000

X, labels = create_gaussian_mixture(n_samples, dim)
plot_data_2D(X, color=labels, node_size=0.5, cmap='coolwarm')
```

```python

exp_params = {
    'p': 3,
}

embedor = EmbedOR(exp_params)
emb = embedor.fit_transform(X)
mds_emb = MDS(n_components=2, random_state=42, dissimilarity='precomputed', verbose=10).fit_transform(embedor.apsp)
tsne_emb = TSNE(n_components=2, random_state=42).fit_transform(X)
umap_emb = umap.UMAP(n_components=2, random_state=42).fit_transform(X)
densemap_emb = umap.UMAP(n_components=2, random_state=42, densmap=True).fit_transform(X)
# mds emb of the embedor apsp
```

```python
# Plotting the embedding
plot_data_2D(X, color=labels, cmap='coolwarm', node_size=0.5)
plt.savefig(os.path.join(output_dir, f'original_data_dim_{dim}.png'), dpi=1200, bbox_inches='tight')
plot_data_2D(emb, color=labels, cmap='coolwarm', node_size=0.5)
plt.savefig(os.path.join(output_dir, f'embedor_embedding_dim_{dim}.png'), dpi=1200, bbox_inches='tight')
plot_data_2D(mds_emb, color=labels, cmap='coolwarm', node_size=0.5)
plt.savefig(os.path.join(output_dir, f'embedor_mds_embedding_dim_{dim}.png'), dpi=1200, bbox_inches='tight')
plot_data_2D(tsne_emb, color=labels, cmap='coolwarm', node_size=0.5)
plt.savefig(os.path.join(output_dir, f'tsne_embedding_dim_{dim}.png'), dpi=1200, bbox_inches='tight')
plot_data_2D(umap_emb, color=labels, cmap='coolwarm', node_size=0.5)
plt.savefig(os.path.join(output_dir, f'umap_embedding_dim_{dim}.png'), dpi=1200, bbox_inches='tight')
plot_data_2D(densemap_emb, color=labels, cmap='coolwarm', node_size=0.5)
plt.savefig(os.path.join(output_dir, f'densemap_embedding_dim_{dim}.png'), dpi=1200, bbox_inches='tight')
# plt.title("Embedding of Gaussian Mixture")
```
